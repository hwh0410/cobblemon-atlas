# Cobblemon Atlas｜烈空座 WebP 播放品質稽核

直接檢查安全鏡頭 V2 的 WebP，不必解壓資源包或重新渲染。檢查動畫時長、解碼影格、重複影格合併與畫面中心跳幅；結果不發布。

```python
from pathlib import Path
import subprocess,sys,json
from google.colab import drive
drive.mount('/content/drive')
ROOT=Path('/content/drive/MyDrive/Cobblemon_Atlas')
SITE=Path('/content/cobblemon-atlas')
BRANCH='feat/animation-manifest-scanner-phase1'
assert ROOT.is_dir()
```

```python
if not (SITE/'.git').is_dir():
    subprocess.run(['git','clone','--depth','1','--filter=blob:none','--sparse','--branch',BRANCH,'https://github.com/hwh0410/cobblemon-atlas.git',str(SITE)],check=True)
subprocess.run(['git','-C',str(SITE),'sparse-checkout','set','tools','notebooks'],check=True)
subprocess.run(['git','-C',str(SITE),'pull','--ff-only','origin',BRANCH],check=True)
print('Git:',subprocess.check_output(['git','-C',str(SITE),'rev-parse','--short','HEAD'],text=True).strip())
```

```python
source=ROOT/'reports/resolver-validation/safe-camera-pilot'
out=source/'webp-playback-audit.json'
assert (source/'0384_rayquaza/physical_attack_safe.webp').is_file()
assert (source/'0384_rayquaza/special_attack_safe.webp').is_file()
subprocess.run([sys.executable,str(SITE/'tools/audit_safe_webp_playback.py'),'--input',str(source),'--out',str(out)],check=True)
report=json.loads(out.read_text())
for name,item in report['files'].items():
    print(name,item.get('quality_status'), 'frames',item.get('decoded_frames'),'duration ms',item.get('decoded_duration_ms'),'warnings',item.get('warnings'))
print('報告:',out)
```
