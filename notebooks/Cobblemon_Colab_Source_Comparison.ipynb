# Cobblemon Atlas｜超夢・烈空座動畫來源比對

直接比對 Drive 的 `merged_pack.zip` 與 `development_resource_packs.zip`，使用 Resolver 已選出的模型骨骼。僅讀動畫 JSON，不解壓整包、不渲染圖片、不覆蓋既有 WebP。

輸出：`Cobblemon_Atlas/reports/resolver-validation/animation-source-comparison.json`。`potential_dynamic_candidate` 代表值得下一輪試渲染，**不是**遊戲語義正確或可發佈。


```python
from pathlib import Path
import subprocess,sys,json
from google.colab import drive
drive.mount('/content/drive')
root=Path('/content/drive/MyDrive/Cobblemon_Atlas')
merged=root/'source_packs/merged_pack.zip'
development=root/'source_packs/development_resource_packs.zip'
resolved=root/'reports/resolver-validation/resolved-forms.json'
for path in (merged,development,resolved):
    assert path.is_file(),f'找不到必要檔案：{path}'
print('ZIP + Resolver inputs verified')
```

```python
branch='feat/animation-manifest-scanner-phase1'
repo=Path('/content/cobblemon-atlas-source-audit')
if not (repo/'.git').is_dir():
    subprocess.run(['git','clone','--depth','1','--filter=blob:none','--sparse','--branch',branch,
                    'https://github.com/hwh0410/cobblemon-atlas.git',str(repo)],check=True)
subprocess.run(['git','-C',str(repo),'sparse-checkout','set','tools'],check=True)
subprocess.run(['git','-C',str(repo),'pull','--ff-only','origin',branch],check=True)
target=root/'reports/resolver-validation/animation-source-comparison.json'
command=[sys.executable,str(repo/'tools/compare_animation_sources.py'),
         '--merged-zip',str(merged),'--development-zip',str(development),
         '--resolved',str(resolved),'--forms','0150_mewtwo','0384_rayquaza','--out',str(target)]
subprocess.run(command,check=True)
print('Saved:',target)
```
